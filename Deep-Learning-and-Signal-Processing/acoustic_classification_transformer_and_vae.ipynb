```python
# OCEAN SOUND CLASSIFICATION
# Assignment 5 - Noah Russell
# CAP 6673

import os
import numpy as np
import pandas as pd
import librosa
import librosa.display
import soundfile as sf
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_recall_fscore_support,
    roc_curve,
    auc,
    roc_auc_score
)
from sklearn.manifold import TSNE
from sklearn.utils.class_weight import compute_class_weight
import warnings
warnings.filterwarnings('ignore')

# Deep Learning imports
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import tensorflow.keras.backend as K

# Progress bar
from tqdm import tqdm
import time

# Set random seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# Plot style
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
print("TensorFlow version:", tf.__version__)

# ==================== GLOBAL CONFIG ====================
SR = 16000          # sample rate
DURATION = 2.0      # seconds per clip
N_MELS = 64         # mel bands
N_FFT = 1024
HOP_LENGTH = 256

# ==================== 1. MOUNT GOOGLE DRIVE & LOAD DATA ====================
print("="*60)
print("1. MOUNTING GOOGLE DRIVE & LOADING DATA")
print("="*60)

from google.colab import drive
drive.mount('/content/drive', force_remount=False)

print("\nYOUR FILE STRUCTURE")
print("Audio files: /content/drive/MyDrive/audios/")
print("Labels file: /content/drive/MyDrive/labels.csv")

labels_path = '/content/drive/MyDrive/labels.csv'
audio_dir = '/content/drive/MyDrive/audios'

print(f"\nChecking files...")
print(f"labels.csv exists: {os.path.exists(labels_path)}")
print(f"audios folder exists: {os.path.exists(audio_dir)}")

if not os.path.exists(labels_path) or not os.path.exists(audio_dir):
    print("ERROR: Files not found! Please check your Google Drive paths.")
    raise SystemExit

# ==================== 2. LOAD AND PREPROCESS DATASET ====================

def load_and_preprocess_dataset():
    """Load label CSV and match with audio files."""
    print("\n" + "="*60)
    print("2. LOADING AND PREPROCESSING DATASET (LABELS)")
    print("="*60)

    try:
        label_df = pd.read_csv(labels_path)
        print(f"✓ Loaded {len(label_df)} entries from labels.csv")

        audio_files = []
        for file in os.listdir(audio_dir):
            if file.lower().endswith('.wav'):
                audio_files.append(os.path.join(audio_dir, file))

        print(f"✓ Found {len(audio_files)} audio files")

        # Create mapping from filename to species
        label_dict = {}
        for _, row in label_df.iterrows():
            filename = str(row['file'])
            filename = os.path.basename(filename)
            species = row['species']
            label_dict[filename] = species

        audio_files.sort()
        labels = []
        matched_audio_files = []

        for audio_path in audio_files:
            filename = os.path.basename(audio_path)
            if filename in label_dict:
                matched_audio_files.append(audio_path)
                labels.append(label_dict[filename])

        print(f"✓ Matched {len(matched_audio_files)} files with labels")

        if len(matched_audio_files) == 0:
            print("ERROR: No files matched!")
            return None, None, None, None

        # Encode labels
        label_encoder = LabelEncoder()
        y = label_encoder.fit_transform(labels)

        print(f"\n✓ Number of classes: {len(label_encoder.classes_)}")
        print(f"✓ Classes: {label_encoder.classes_}")

        # Show class distribution
        print("\nClass distribution:")
        unique, counts = np.unique(y, return_counts=True)
        for cls, count in zip(unique, counts):
            class_name = label_encoder.inverse_transform([cls])[0]
            print(f"  {class_name}: {count} samples")

        return matched_audio_files, labels, y, label_encoder

    except Exception as e:
        print(f"ERROR in load_and_preprocess_dataset: {e}")
        import traceback
        traceback.print_exc()
        return None, None, None, None

# Load dataset
audio_files, label_names, y, label_encoder = load_and_preprocess_dataset()

if audio_files is None:
    print("Failed to load dataset. Exiting.")
    raise SystemExit

# ==================== 3. OPTIMIZED AUDIO PREPROCESSING ====================

def load_audio_file(file_path, target_sr=SR, duration=DURATION):
    """Load and preprocess single audio file efficiently."""
    try:
        audio, sr = librosa.load(file_path, sr=target_sr, mono=True, duration=duration)

        target_samples = int(duration * target_sr)
        if len(audio) > target_samples:
            audio = audio[:target_samples]
        else:
            padding = target_samples - len(audio)
            audio = np.pad(audio, (0, padding), mode='constant')

        # Normalize
        max_val = np.max(np.abs(audio))
        if max_val > 0:
            audio = audio / (max_val + 1e-8)

        return audio.astype(np.float32)
    except Exception as e:
        print(f"Error loading {file_path}: {e}")
        return None

def create_mel_spectrogram(audio, sr=SR, n_mels=N_MELS, n_fft=N_FFT, hop_length=HOP_LENGTH):
    """Create log-mel spectrogram normalized to [0,1]."""
    if np.max(np.abs(audio)) < 1e-6:
        audio = np.random.normal(0, 0.01, len(audio))

    mel_spec = librosa.feature.melspectrogram(
        y=audio,
        sr=sr,
        n_mels=n_mels,
        n_fft=n_fft,
        hop_length=hop_length,
        fmin=20,
        fmax=sr // 2
    )
    mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)

    # Normalize to [0, 1] using a fixed dB range [-80, 0]
    mel_spec_db = np.clip(mel_spec_db, -80.0, 0.0)
    mel_norm = (mel_spec_db + 80.0) / 80.0  # [-80,0] -> [0,1]

    return mel_norm.astype(np.float32)

def preprocess_all_files(audio_files, y, target_sr=SR, duration=DURATION):
    """Preprocess all audio files into waveforms + spectrograms."""
    print(f"\nPreprocessing {len(audio_files)} audio files...")

    X_audio = []
    X_spec = []
    valid_indices = []

    for i, file_path in enumerate(tqdm(audio_files, desc="Processing audio files")):
        audio = load_audio_file(file_path, target_sr, duration)
        if audio is not None and len(audio) > 0:
            try:
                spec = create_mel_spectrogram(audio, target_sr)
                if spec is not None and spec.shape[0] > 0 and spec.shape[1] > 0:
                    X_audio.append(audio)
                    X_spec.append(spec)
                    valid_indices.append(i)
            except Exception as e:
                print(f"Error creating spectrogram for {file_path}: {e}")

    if len(X_audio) == 0:
        print("ERROR: No audio files processed successfully!")
        return None, None, None

    X_audio = np.array(X_audio, dtype=np.float32)
    X_spec = np.array(X_spec, dtype=np.float32)
    y_valid = np.array([y[i] for i in valid_indices])

    print(f"\n✓ Successfully processed {len(X_audio)} files")
    print(f"✓ Audio shape: {X_audio.shape}")
    print(f"✓ Spectrogram shape: {X_spec.shape}  (n_mels={X_spec.shape[1]}, T={X_spec.shape[2]})")

    return X_audio, X_spec, y_valid

# Preprocess all files
X_audio, X_spec, y = preprocess_all_files(audio_files, y)

if X_audio is None:
    print("Failed to preprocess files. Exiting.")
    raise SystemExit

# ==================== 4. CREATE DATA SPLITS ====================
print("\n" + "="*60)
print("3. CREATING DATA SPLITS")
print("="*60)

# Add channel dimension for spectrograms: [N, H, W, 1]
X_spec = X_spec[..., np.newaxis]
print(f"Final spectrogram shape (with channel): {X_spec.shape}")

# Split spectrogram data
X_spec_train_full, X_spec_test, y_spec_train_full, y_spec_test = train_test_split(
    X_spec, y, test_size=0.2, stratify=y, random_state=42
)

X_spec_train, X_spec_val, y_spec_train, y_spec_val = train_test_split(
    X_spec_train_full, y_spec_train_full, test_size=0.125,
    stratify=y_spec_train_full, random_state=42
)

# Split raw audio data
X_audio = X_audio[..., np.newaxis]  # [N, T, 1]
print(f"Raw audio shape before split: {X_audio.shape}")

X_audio_train_full, X_audio_test, y_audio_train_full, y_audio_test = train_test_split(
    X_audio, y, test_size=0.2, stratify=y, random_state=42
)

X_audio_train, X_audio_val, y_audio_train, y_audio_val = train_test_split(
    X_audio_train_full, y_audio_train_full, test_size=0.125,
    stratify=y_audio_train_full, random_state=42
)

print(f"\n✓ Spectrogram data:")
print(f"  Train: {X_spec_train.shape[0]} samples, shape: {X_spec_train.shape[1:]}")
print(f"  Validation: {X_spec_val.shape[0]} samples")
print(f"  Test: {X_spec_test.shape[0]} samples")

print(f"\n✓ Raw audio data:")
print(f"  Train: {X_audio_train.shape[0]} samples, shape: {X_audio_train.shape[1:]}")
print(f"  Validation: {X_audio_val.shape[0]} samples")
print(f"  Test: {X_audio_test.shape[0]} samples")

# ==================== 5. VISUALIZATION ====================
print("\n" + "="*60)
print("4. DATASET VISUALIZATION")
print("="*60)

def plot_spectrogram_samples(X, y, label_encoder, n_classes=5, n_samples=2):
    """Plot spectrogram samples from different classes."""
    unique_classes = np.unique(y)
    if len(unique_classes) < n_classes:
        n_classes = len(unique_classes)

    fig, axes = plt.subplots(
        n_samples, n_classes,
        figsize=(4 * n_classes, 3 * n_samples),
        squeeze=False,
        constrained_layout=True
    )

    for i, cls in enumerate(unique_classes[:n_classes]):
        class_indices = np.where(y == cls)[0]
        if len(class_indices) > 0:
            for j in range(min(n_samples, len(class_indices))):
                idx = class_indices[j]
                ax = axes[j, i]

                if len(X.shape) == 4:
                    spec = X[idx, :, :, 0]
                else:
                    spec = X[idx]

                img = librosa.display.specshow(
                    spec, ax=ax, cmap='viridis',
                    x_axis=None, y_axis=None
                )
                ax.set_xticks([])
                ax.set_yticks([])
                if j == 0:
                    class_name = label_encoder.inverse_transform([cls])[0]
                    ax.set_title(f"{class_name}", fontsize=10, pad=10)

    fig.colorbar(
        img, ax=axes.ravel().tolist(), orientation='vertical',
        fraction=0.02, pad=0.04, label='Normalized Power'
    )

    plt.suptitle('Sample Spectrograms from Different Marine Mammal Species',
                 fontsize=14, y=1.02)
    plt.show()

# Plot the samples
if len(X_spec_train) > 0:
    plot_spectrogram_samples(
        X_spec_train, y_spec_train, label_encoder,
        n_classes=min(3, len(label_encoder.classes_)),
        n_samples=2
    )

# Plot class distribution
plt.figure(figsize=(12, 6))
class_counts = pd.Series(y_spec_train).value_counts().sort_index()
class_names = [label_encoder.inverse_transform([i])[0] for i in class_counts.index]
bars = plt.bar(range(len(class_counts)), class_counts.values)
plt.xlabel('Class Index', fontsize=12)
plt.ylabel('Number of Samples', fontsize=12)
plt.title('Class Distribution in Training Set', fontsize=14)
plt.xticks(
    range(len(class_counts)),
    [f"{idx}" for idx in class_counts.index],
    rotation=90
)
plt.grid(axis='y', alpha=0.3)

for bar, count in zip(bars, class_counts.values):
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2., height + 0.5,
        f'{count}', ha='center', va='bottom', fontsize=9
    )

plt.tight_layout()
plt.show()

print(f"\nTraining set has {len(np.unique(y_spec_train))} classes with {len(X_spec_train)} total samples")
print(f"Class with most samples: {class_names[0]} ({class_counts.iloc[0]} samples)")
print(f"Class with fewest samples: {class_names[-1]} ({class_counts.iloc[-1]} samples)")

# Compute class weights
class_weights_array = compute_class_weight('balanced', classes=np.unique(y_spec_train), y=y_spec_train)
class_weights = {i: float(class_weights_array[i]) for i in range(len(class_weights_array))}
print(f"\nClass weights for handling imbalance: {class_weights}")

# ==================== 6. COMMON UTILITIES ====================

def specs_to_sequences(X):
    """Convert spectrogram images [B, H, W, 1] to sequences [B, T, F]."""
    return np.transpose(X, (0, 2, 1, 3)).reshape(X.shape[0], X.shape[2], X.shape[1])

def spec_image_to_audio(spec_img, sr=SR, n_fft=N_FFT, hop_length=HOP_LENGTH, n_iter=32):
    """Convert spectrogram back to audio waveform."""
    if spec_img.ndim == 3:
        spec_img = spec_img[:, :, 0]

    # Undo [0,1] -> [-80,0] dB
    mel_db = spec_img * 80.0 - 80.0
    mel_power = librosa.db_to_power(mel_db)

    # Invert mel spectrogram to waveform
    audio = librosa.feature.inverse.mel_to_audio(
        mel_power,
        sr=sr,
        n_fft=n_fft,
        hop_length=hop_length,
        n_iter=n_iter
    )
    max_val = np.max(np.abs(audio))
    if max_val > 0:
        audio = audio / (max_val + 1e-8)
    return audio

class Timer:
    """Context manager for timing code execution."""
    def __enter__(self):
        self.start = time.time()
        return self

    def __exit__(self, *args):
        self.end = time.time()
        self.duration = self.end - self.start
        print(f"Execution time: {self.duration:.2f} seconds")

class ModelTrainer:
    """Utility class for training and evaluating models."""

    def __init__(self, label_encoder):
        self.label_encoder = label_encoder
        self.results = {}
        self.histories = {}

    def train_model(self, model, X_train, y_train, X_val, y_val, model_name, epochs=30, class_weight=None):
        """Train a model with early stopping and LR scheduling."""
        print(f"\nTraining {model_name}...")
        print(f"Training on {X_train.shape[0]} samples, validating on {X_val.shape[0]} samples")

        callbacks = [
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=10,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            )
        ]

        model.compile(
            optimizer=keras.optimizers.Adam(learning_rate=0.001),
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy']
        )

        with Timer():
            history = model.fit(
                X_train, y_train,
                validation_data=(X_val, y_val),
                epochs=epochs,
                batch_size=32,
                callbacks=callbacks,
                verbose=1,
                class_weight=class_weight
            )

        self.histories[model_name] = history
        return history, model

    def evaluate_model(self, model, X_test, y_test, model_name):
        """Evaluate a model and store metrics."""
        print(f"\nEvaluating {model_name}...")
        print(f"Testing on {X_test.shape[0]} samples")

        y_pred_proba = model.predict(X_test, verbose=0, batch_size=32)
        y_pred = np.argmax(y_pred_proba, axis=1)

        accuracy = accuracy_score(y_test, y_pred)
        precision, recall, f1, _ = precision_recall_fscore_support(
            y_test, y_pred, average='weighted'
        )

        # Calculate ROC AUC for multi-class
        try:
            roc_auc = roc_auc_score(y_test, y_pred_proba, multi_class='ovr', average='weighted')
        except:
            roc_auc = 0.0

        self.results[model_name] = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1_score': f1,
            'roc_auc': roc_auc,
            'predictions': y_pred,
            'probabilities': y_pred_proba,
            'model': model
        }

        print(f"✓ Accuracy: {accuracy:.4f}")
        print(f"✓ Precision: {precision:.4f}")
        print(f"✓ Recall: {recall:.4f}")
        print(f"✓ F1-Score: {f1:.4f}")
        print(f"✓ ROC AUC: {roc_auc:.4f}")

        unique_preds = np.unique(y_pred)
        unique_true = np.unique(y_test)

        print(f"\nPredicted {len(unique_preds)} unique classes")
        print(f"True labels have {len(unique_true)} unique classes")

        missing_classes = set(int(c) for c in unique_true) - set(int(c) for c in unique_preds)
        if missing_classes:
            print(f"Warning: Model didn't predict {len(missing_classes)} classes")

        all_labels = np.arange(len(self.label_encoder.classes_))

        print("\nDetailed Classification Report:")
        print(classification_report(
            y_test, y_pred,
            labels=all_labels,
            target_names=self.label_encoder.classes_,
            zero_division=0
        ))

        self.plot_confusion_matrix(y_test, y_pred, model_name)

        return accuracy, precision, recall, f1

    def plot_confusion_matrix(self, y_true, y_pred, model_name):
        """Plot normalized confusion matrix."""
        plt.figure(figsize=(10, 8))
        num_classes = len(self.label_encoder.classes_)
        all_labels = np.arange(num_classes)

        cm = confusion_matrix(y_true, y_pred, labels=all_labels)
        labels = self.label_encoder.classes_

        with np.errstate(divide='ignore', invalid='ignore'):
            cm_normalized = cm.astype('float') / cm.sum(axis=1, keepdims=True)
            cm_normalized = np.nan_to_num(cm_normalized)

        sns.heatmap(
            cm_normalized, annot=False, fmt='.2f', cmap='Blues',
            xticklabels=labels,
            yticklabels=labels,
            cbar_kws={'label': 'Normalized Accuracy'}
        )

        plt.title(f'Confusion Matrix - {model_name}', fontsize=14, pad=20)
        plt.ylabel('True Label', fontsize=12)
        plt.xlabel('Predicted Label', fontsize=12)
        plt.xticks(rotation=90, ha='right')
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()

    def plot_training_history(self, model_name):
        """Plot training history for a specific model."""
        if model_name not in self.histories:
            print(f"No history found for {model_name}")
            return

        history = self.histories[model_name]
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

        # Plot accuracy
        ax1.plot(history.history['accuracy'], label='Train Accuracy')
        ax1.plot(history.history['val_accuracy'], label='Val Accuracy')
        ax1.set_title(f'{model_name} - Accuracy')
        ax1.set_xlabel('Epoch')
        ax1.set_ylabel('Accuracy')
        ax1.legend()
        ax1.grid(True)

        # Plot loss
        ax2.plot(history.history['loss'], label='Train Loss')
        ax2.plot(history.history['val_loss'], label='Val Loss')
        ax2.set_title(f'{model_name} - Loss')
        ax2.set_xlabel('Epoch')
        ax2.set_ylabel('Loss')
        ax2.legend()
        ax2.grid(True)

        plt.tight_layout()
        plt.show()

# Initialize trainer
trainer = ModelTrainer(label_encoder)

# ==================== TASK 1: SPECTROGRAM + DNN ====================
print("\n" + "="*60)
print("TASK 1: SPECTROGRAM + DNN")
print("="*60)

dnn_model = keras.Sequential([
    keras.layers.Input(shape=X_spec_train.shape[1:]),
    keras.layers.Flatten(),
    keras.layers.Dense(512, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

dnn_history, dnn_model = trainer.train_model(
    dnn_model,
    X_spec_train, y_spec_train,
    X_spec_val, y_spec_val,
    model_name="Spectrogram DNN",
    epochs=40,
    class_weight=class_weights
)

dnn_accuracy = trainer.evaluate_model(
    dnn_model,
    X_spec_test, y_spec_test,
    "Spectrogram DNN"
)[0]

trainer.plot_training_history("Spectrogram DNN")

# ==================== TASK 2: SPECTROGRAM + CNN ====================
print("\n" + "="*60)
print("TASK 2: SPECTROGRAM + CNN")
print("="*60)

cnn_model = keras.Sequential([
    keras.layers.Input(shape=X_spec_train.shape[1:]),

    # First conv block
    keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.2),

    # Second conv block
    keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.3),

    # Third conv block
    keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.4),

    # Global pooling and dense layers
    keras.layers.GlobalAveragePooling2D(),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

cnn_history, cnn_model = trainer.train_model(
    cnn_model,
    X_spec_train, y_spec_train,
    X_spec_val, y_spec_val,
    model_name="Spectrogram CNN",
    epochs=40,
    class_weight=class_weights
)

cnn_accuracy = trainer.evaluate_model(
    cnn_model,
    X_spec_test, y_spec_test,
    "Spectrogram CNN"
)[0]

trainer.plot_training_history("Spectrogram CNN")

# ==================== TASK 3: SPECTROGRAM + RNN ====================
print("\n" + "="*60)
print("TASK 3: SPECTROGRAM + RNN")
print("="*60)

# Convert spectrograms to sequences [B, T, F]
X_spec_train_rnn = specs_to_sequences(X_spec_train)
X_spec_val_rnn   = specs_to_sequences(X_spec_val)
X_spec_test_rnn  = specs_to_sequences(X_spec_test)

print("RNN input shape:", X_spec_train_rnn.shape)  # [B, T, F]

rnn_model = keras.Sequential([
    keras.layers.Input(shape=(X_spec_train_rnn.shape[1], X_spec_train_rnn.shape[2])),
    keras.layers.SimpleRNN(256, return_sequences=True, dropout=0.3, recurrent_dropout=0.3),
    keras.layers.BatchNormalization(),
    keras.layers.SimpleRNN(128, dropout=0.3, recurrent_dropout=0.3),
    keras.layers.BatchNormalization(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

rnn_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Training RNN...")
with Timer():
    rnn_history = rnn_model.fit(
        X_spec_train_rnn, y_spec_train,
        validation_data=(X_spec_val_rnn, y_spec_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=10,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            )
        ]
    )

trainer.histories["Spectrogram RNN"] = rnn_history

rnn_accuracy = trainer.evaluate_model(
    rnn_model,
    X_spec_test_rnn, y_spec_test,
    "Spectrogram RNN"
)[0]

# ==================== TASK 4: SPECTROGRAM + LSTM  ====================
print("\n" + "="*60)
print("TASK 4: SPECTROGRAM + LSTM")
print("="*60)

lstm_model = keras.Sequential([
    keras.layers.Input(shape=(X_spec_train_rnn.shape[1], X_spec_train_rnn.shape[2])),
    keras.layers.LSTM(256, return_sequences=True, dropout=0.3, recurrent_dropout=0.3),
    keras.layers.BatchNormalization(),
    keras.layers.LSTM(128, dropout=0.3, recurrent_dropout=0.3),
    keras.layers.BatchNormalization(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dropout(0.3),
    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

lstm_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Training LSTM...")
with Timer():
    lstm_history = lstm_model.fit(
        X_spec_train_rnn, y_spec_train,
        validation_data=(X_spec_val_rnn, y_spec_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=10,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            )
        ]
    )

trainer.histories["Spectrogram LSTM"] = lstm_history

lstm_accuracy = trainer.evaluate_model(
    lstm_model,
    X_spec_test_rnn, y_spec_test,
    "Spectrogram LSTM"
)[0]

# ==================== TASK 5: SPECTROGRAM + TRANSFORMER ====================
print("\n" + "="*60)
print("TASK 5: SPECTROGRAM + TRANSFORMER")
print("="*60)

d_model = 128
n_heads = 8

transformer_input = keras.layers.Input(shape=(X_spec_train_rnn.shape[1], X_spec_train_rnn.shape[2]))

# Project features to d_model
x = keras.layers.Dense(d_model)(transformer_input)

# Positional encoding
positions = tf.range(start=0, limit=X_spec_train_rnn.shape[1], delta=1)
pos_encoding = keras.layers.Embedding(
    input_dim=X_spec_train_rnn.shape[1],
    output_dim=d_model
)(positions)
pos_encoding = tf.expand_dims(pos_encoding, axis=0)
x = x + pos_encoding

# Transformer encoder blocks
for _ in range(3):
    # Multi-head attention
    attn_output = keras.layers.MultiHeadAttention(
        num_heads=n_heads,
        key_dim=d_model // n_heads,
        dropout=0.1
    )(x, x)
    attn_output = keras.layers.Dropout(0.1)(attn_output)
    x = keras.layers.LayerNormalization()(x + attn_output)

    # Feed-forward network
    ffn = keras.layers.Dense(4 * d_model, activation='relu')(x)
    ffn = keras.layers.Dropout(0.1)(ffn)
    ffn = keras.layers.Dense(d_model)(ffn)
    ffn = keras.layers.Dropout(0.1)(ffn)
    x = keras.layers.LayerNormalization()(x + ffn)

# Pooling and classification
pooled = keras.layers.GlobalAveragePooling1D()(x)
dropout = keras.layers.Dropout(0.5)(pooled)
dense1 = keras.layers.Dense(256, activation='relu')(dropout)
dropout2 = keras.layers.Dropout(0.4)(dense1)
dense2 = keras.layers.Dense(128, activation='relu')(dropout2)
dropout3 = keras.layers.Dropout(0.3)(dense2)
outputs = keras.layers.Dense(len(label_encoder.classes_), activation='softmax')(dropout3)

transformer_model = keras.Model(inputs=transformer_input, outputs=outputs)

transformer_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Training Transformer...")
with Timer():
    transformer_history = transformer_model.fit(
        X_spec_train_rnn, y_spec_train,
        validation_data=(X_spec_val_rnn, y_spec_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=10,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            )
        ]
    )

trainer.histories["Spectrogram Transformer"] = transformer_history

transformer_accuracy = trainer.evaluate_model(
    transformer_model,
    X_spec_test_rnn, y_spec_test,
    "Spectrogram Transformer"
)[0]

# ==================== TASK 6: RAW AUDIO CLASSIFICATION ====================
print("\n" + "="*60)
print("TASK 6: RAW AUDIO CLASSIFICATION")
print("="*60)

raw_audio_model = keras.Sequential([
    keras.layers.Input(shape=X_audio_train.shape[1:]),   # [T, 1]

    # First conv block
    keras.layers.Conv1D(64, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv1D(64, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling1D(2),
    keras.layers.Dropout(0.3),

    # Second conv block
    keras.layers.Conv1D(128, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv1D(128, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling1D(2),
    keras.layers.Dropout(0.3),

    # Third conv block
    keras.layers.Conv1D(256, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.Conv1D(256, 5, activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling1D(2),
    keras.layers.Dropout(0.4),

    # Global pooling and dense layers
    keras.layers.GlobalAveragePooling1D(),
    keras.layers.Dense(512, activation='relu'),
    keras.layers.BatchNormalization(),
    keras.layers.Dropout(0.5),
    keras.layers.Dense(256, activation='relu'),
    keras.layers.Dropout(0.4),
    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

raw_audio_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Training Raw Audio 1D-CNN...")
with Timer():
    raw_audio_history = raw_audio_model.fit(
        X_audio_train, y_audio_train,
        validation_data=(X_audio_val, y_audio_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=10,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=5,
                min_lr=1e-6,
                verbose=1
            )
        ]
    )

trainer.histories["Raw Audio 1D-CNN"] = raw_audio_history

raw_audio_accuracy = trainer.evaluate_model(
    raw_audio_model,
    X_audio_test, y_audio_test,
    "Raw Audio 1D-CNN"
)[0]

# ==================== TASK 7: CONTRASTIVE LEARNING  ====================
print("\n" + "="*60)
print("TASK 7: CONTRASTIVE LEARNING (Simplified)")
print("="*60)

def build_simple_contrastive_model(input_shape, embedding_dim=128):
    """Build a simple contrastive learning encoder."""
    inputs = keras.layers.Input(shape=input_shape)

    # Feature extractor
    x = keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same')(inputs)
    x = keras.layers.BatchNormalization()(x)
    x = keras.layers.MaxPooling2D((2, 2))(x)

    x = keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same')(x)
    x = keras.layers.BatchNormalization()(x)
    x = keras.layers.MaxPooling2D((2, 2))(x)

    x = keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same')(x)
    x = keras.layers.BatchNormalization()(x)
    x = keras.layers.GlobalAveragePooling2D()(x)

    # Embedding layer
    embeddings = keras.layers.Dense(embedding_dim, activation='linear')(x)

    return keras.Model(inputs=inputs, outputs=embeddings, name="contrastive_encoder")

# Build contrastive encoder
contrastive_encoder = build_simple_contrastive_model(X_spec_train.shape[1:], embedding_dim=128)

# Build classification head
contrastive_input = keras.layers.Input(shape=X_spec_train.shape[1:])
embeddings = contrastive_encoder(contrastive_input)
x = keras.layers.Dense(256, activation='relu')(embeddings)
x = keras.layers.BatchNormalization()(x)
x = keras.layers.Dropout(0.5)(x)
x = keras.layers.Dense(128, activation='relu')(x)
x = keras.layers.Dropout(0.4)(x)
contrastive_output = keras.layers.Dense(len(label_encoder.classes_), activation='softmax')(x)

contrastive_model = keras.Model(inputs=contrastive_input, outputs=contrastive_output, name="contrastive_classifier")

# Compile with a higher learning rate
contrastive_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0005),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Training contrastive model...")
with Timer():
    contrastive_history = contrastive_model.fit(
        X_spec_train, y_spec_train,
        validation_data=(X_spec_val, y_spec_val),
        epochs=50,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=15,
                restore_best_weights=True,
                verbose=1
            ),
            keras.callbacks.ReduceLROnPlateau(
                monitor='val_loss',
                factor=0.5,
                patience=8,
                min_lr=1e-6,
                verbose=1
            )
        ]
    )

trainer.histories["Contrastive Learning"] = contrastive_history

contrastive_accuracy = trainer.evaluate_model(
    contrastive_model,
    X_spec_test, y_spec_test,
    "Contrastive Learning"
)[0]

# Visualize embeddings ( move to CPU properly)
print("\nVisualizing contrastive embeddings with t-SNE...")
try:
    # Create a separate model just for embeddings
    embedding_model = keras.Model(inputs=contrastive_encoder.input, outputs=contrastive_encoder.output)

    # Get embeddings on a subset
    n_samples = min(200, len(X_spec_test))
    embeddings_test = embedding_model.predict(X_spec_test[:n_samples], verbose=0)

    # Use t-SNE
    tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
    embeddings_2d = tsne.fit_transform(embeddings_test)

    plt.figure(figsize=(12, 10))
    scatter = plt.scatter(
        embeddings_2d[:, 0], embeddings_2d[:, 1],
        c=y_spec_test[:n_samples], cmap='tab20c', alpha=0.7, s=50
    )
    plt.colorbar(scatter)
    plt.title('Contrastive Learning Embeddings (t-SNE)', fontsize=14)
    plt.xlabel('t-SNE 1')
    plt.ylabel('t-SNE 2')
    plt.grid(alpha=0.3)
    plt.show()

except Exception as e:
    print(f"Could not visualize embeddings: {e}")

# ==================== TASK 8: VAE FOR DATA AUGMENTATION  ====================
print("\n" + "="*60)
print("TASK 8: VAE FOR DATA AUGMENTATION")
print("="*60)

class VAE(keras.Model):
    """Variational Autoencoder."""

    def __init__(self, input_shape, latent_dim=32):
        super(VAE, self).__init__()
        self.latent_dim = latent_dim
        self.input_shape_val = input_shape

        # Encoder
        encoder_inputs = keras.layers.Input(shape=input_shape)
        x = keras.layers.Flatten()(encoder_inputs)
        x = keras.layers.Dense(512, activation='relu')(x)
        x = keras.layers.Dense(256, activation='relu')(x)
        x = keras.layers.Dense(128, activation='relu')(x)

        self.z_mean = keras.layers.Dense(latent_dim)(x)
        self.z_log_var = keras.layers.Dense(latent_dim)(x)

        self.encoder = keras.Model(encoder_inputs, [self.z_mean, self.z_log_var], name="encoder")

        # Decoder
        latent_inputs = keras.layers.Input(shape=(latent_dim,))
        x = keras.layers.Dense(128, activation='relu')(latent_inputs)
        x = keras.layers.Dense(256, activation='relu')(x)
        x = keras.layers.Dense(512, activation='relu')(x)
        x = keras.layers.Dense(np.prod(input_shape), activation='sigmoid')(x)
        decoder_outputs = keras.layers.Reshape(input_shape)(x)

        self.decoder = keras.Model(latent_inputs, decoder_outputs, name="decoder")

    def sample(self, z_mean, z_log_var):
        """Sample from the latent distribution."""
        batch = tf.shape(z_mean)[0]
        dim = tf.shape(z_mean)[1]
        epsilon = tf.random.normal(shape=(batch, dim))
        return z_mean + tf.exp(0.5 * z_log_var) * epsilon

    def call(self, inputs):
        """Forward pass."""
        z_mean, z_log_var = self.encoder(inputs)
        z = self.sample(z_mean, z_log_var)
        reconstructed = self.decoder(z)

        # Add KL divergence loss
        kl_loss = -0.5 * tf.reduce_mean(
            tf.reduce_sum(1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var), axis=1)
        )

        # Add reconstruction loss
        reconstruction_loss = tf.reduce_mean(
            tf.reduce_sum(
                keras.losses.binary_crossentropy(inputs, reconstructed),
                axis=(1, 2)
            )
        )

        self.add_loss(kl_loss + reconstruction_loss)

        return reconstructed

# Create and compile VAE
vae = VAE(X_spec_train.shape[1:], latent_dim=32)
vae.compile(optimizer=keras.optimizers.Adam(learning_rate=0.001))

print("Training VAE...")
X_spec_train_normalized = X_spec_train.astype('float32')

with Timer():
    vae.fit(
        X_spec_train_normalized, X_spec_train_normalized,
        epochs=30,
        batch_size=32,
        verbose=1
    )

# Generate synthetic samples
print("\nGenerating synthetic samples with VAE...")
synthetic_samples = []
synthetic_labels = []

unique_classes, counts = np.unique(y_spec_train, return_counts=True)
class_to_count = dict(zip(unique_classes, counts))

for class_idx, count in class_to_count.items():
    n_new = max(5, int(0.3 * count))  # 30% extra per class
    for _ in range(n_new):
        # Sample from latent space
        z = np.random.normal(0, 1, size=(1, 32))
        generated = vae.decoder.predict(z, verbose=0)
        synthetic_samples.append(generated[0])
        synthetic_labels.append(class_idx)

synthetic_samples = np.array(synthetic_samples, dtype=np.float32)
synthetic_labels = np.array(synthetic_labels, dtype=np.int32)

print(f"Generated {len(synthetic_samples)} synthetic samples")

# Visualize VAE-generated samples
print("\nVisualizing VAE-generated samples...")
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, ax in enumerate(axes.flat):
    if i < min(10, len(synthetic_samples)):
        ax.imshow(synthetic_samples[i, :, :, 0], cmap='viridis', aspect='auto')
        class_idx = synthetic_labels[i]
        class_name = label_encoder.inverse_transform([class_idx])[0]
        ax.set_title(f"{class_name[:15]}...")
        ax.axis('off')
plt.suptitle('VAE Generated Spectrograms', fontsize=14)
plt.tight_layout()
plt.show()

# Train CNN with VAE augmentation
X_spec_augmented = np.concatenate([X_spec_train, synthetic_samples], axis=0)
y_spec_augmented = np.concatenate([y_spec_train, synthetic_labels], axis=0)

print(f"\nOriginal samples: {len(X_spec_train)}")
print(f"VAE synthetic samples: {len(synthetic_samples)}")
print(f"Total augmented samples: {len(X_spec_augmented)}")

# Create a simpler CNN for augmented training
cnn_vae_model = keras.Sequential([
    keras.layers.Input(shape=X_spec_train.shape[1:]),
    keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.3),

    keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.4),

    keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    keras.layers.GlobalAveragePooling2D(),
    keras.layers.Dropout(0.5),

    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.4),

    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

cnn_vae_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0005),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nTraining CNN with VAE-augmented data...")
with Timer():
    cnn_vae_history = cnn_vae_model.fit(
        X_spec_augmented, y_spec_augmented,
        validation_data=(X_spec_val, y_spec_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=15,
                restore_best_weights=True,
                verbose=1
            )
        ]
    )

trainer.histories["CNN with VAE Augmentation"] = cnn_vae_history

cnn_vae_accuracy = trainer.evaluate_model(
    cnn_vae_model,
    X_spec_test, y_spec_test,
    "CNN with VAE Augmentation"
)[0]

# ==================== TASK 9: SIMPLIFIED DATA AUGMENTATION ====================
print("\n" + "="*60)
print("TASK 9: SIMPLIFIED DATA AUGMENTATION (MixUp)")
print("="*60)

print("Using MixUp data augmentation technique...")

def mixup_data(X, y, alpha=0.2):
    """Apply MixUp augmentation."""
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1

    batch_size = X.shape[0]
    index = np.random.permutation(batch_size)

    mixed_x = lam * X + (1 - lam) * X[index]
    mixed_y = lam * y + (1 - lam) * y[index]

    return mixed_x, mixed_y

# Create augmented dataset using MixUp
X_mixup = []
y_mixup = []

# Generate MixUp samples for each class
for class_idx in np.unique(y_spec_train):
    class_indices = np.where(y_spec_train == class_idx)[0]
    if len(class_indices) > 1:
        n_mixup = min(20, len(class_indices) * 2)  # Generate some mixup samples
        for _ in range(n_mixup):
            idx1, idx2 = np.random.choice(class_indices, 2, replace=False)
            lam = np.random.beta(0.2, 0.2)
            mixed_x = lam * X_spec_train[idx1] + (1 - lam) * X_spec_train[idx2]
            mixed_y = np.zeros(len(label_encoder.classes_))
            mixed_y[class_idx] = 1.0  # Keep the same class label

            X_mixup.append(mixed_x)
            y_mixup.append(class_idx)

X_mixup = np.array(X_mixup, dtype=np.float32)
y_mixup = np.array(y_mixup, dtype=np.int32)

print(f"Generated {len(X_mixup)} MixUp samples")

# Combine with original data
X_spec_mixup_aug = np.concatenate([X_spec_train, X_mixup], axis=0)
y_spec_mixup_aug = np.concatenate([y_spec_train, y_mixup], axis=0)

print(f"\nOriginal samples: {len(X_spec_train)}")
print(f"MixUp samples: {len(X_mixup)}")
print(f"Total augmented samples: {len(X_spec_mixup_aug)}")

# Train model with MixUp augmentation
cnn_mixup_model = keras.Sequential([
    keras.layers.Input(shape=X_spec_train.shape[1:]),
    keras.layers.Conv2D(32, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.3),

    keras.layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
    keras.layers.BatchNormalization(),
    keras.layers.MaxPooling2D((2, 2)),
    keras.layers.Dropout(0.4),

    keras.layers.Conv2D(128, (3, 3), activation='relu', padding='same'),
    keras.layers.GlobalAveragePooling2D(),
    keras.layers.Dropout(0.5),

    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.4),

    keras.layers.Dense(len(label_encoder.classes_), activation='softmax')
])

cnn_mixup_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.0005),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("\nTraining CNN with MixUp augmentation...")
with Timer():
    cnn_mixup_history = cnn_mixup_model.fit(
        X_spec_mixup_aug, y_spec_mixup_aug,
        validation_data=(X_spec_val, y_spec_val),
        epochs=40,
        batch_size=32,
        verbose=1,
        class_weight=class_weights,
        callbacks=[
            keras.callbacks.EarlyStopping(
                monitor='val_loss',
                patience=15,
                restore_best_weights=True,
                verbose=1
            )
        ]
    )

trainer.histories["CNN with MixUp Augmentation"] = cnn_mixup_history

cnn_mixup_accuracy = trainer.evaluate_model(
    cnn_mixup_model,
    X_spec_test, y_spec_test,
    "CNN with MixUp Augmentation"
)[0]

# ==================== COMPREHENSIVE RESULTS ANALYSIS ====================
print("\n" + "="*80)
print("COMPREHENSIVE RESULTS ANALYSIS")
print("="*80)

# Create comparison table
all_results = []
for model_name, results in trainer.results.items():
    if 'accuracy' in results:
        all_results.append({
            'Model': model_name,
            'Accuracy': results['accuracy'],
            'Precision': results['precision'],
            'Recall': results['recall'],
            'F1-Score': results['f1_score'],
            'ROC AUC': results.get('roc_auc', 0.0)
        })

results_df = pd.DataFrame(all_results)
results_df = results_df.sort_values('Accuracy', ascending=False)

print("\n" + "="*80)
print("PERFORMANCE COMPARISON (Ranked by Accuracy)")
print("="*80)
print(results_df.to_string(index=False, float_format=lambda x: f'{x:.4f}'))

# Plot comparison
plt.figure(figsize=(14, 8))
x_pos = np.arange(len(results_df))
width = 0.15

fig, ax = plt.subplots(figsize=(16, 8))
x = np.arange(len(results_df))
width = 0.15

ax.bar(x - 2*width, results_df['Accuracy'], width, label='Accuracy', color='skyblue')
ax.bar(x - width, results_df['Precision'], width, label='Precision', color='lightcoral')
ax.bar(x, results_df['Recall'], width, label='Recall', color='lightgreen')
ax.bar(x + width, results_df['F1-Score'], width, label='F1-Score', color='gold')
ax.bar(x + 2*width, results_df['ROC AUC'], width, label='ROC AUC', color='violet')

ax.set_xlabel('Model')
ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison', fontsize=16, pad=20)
ax.set_xticks(x)
ax.set_xticklabels(results_df['Model'], rotation=45, ha='right', fontsize=10)
ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Plot accuracy comparison
plt.figure(figsize=(14, 8))
colors = plt.cm.Set2(np.linspace(0, 1, len(results_df)))
bars = plt.barh(range(len(results_df)), results_df['Accuracy'], color=colors)
plt.xlabel('Accuracy', fontsize=12)
plt.title('Model Accuracy Comparison - Ocean Sound Classification', fontsize=14, pad=20)
plt.yticks(range(len(results_df)), results_df['Model'], fontsize=11)
plt.xlim(0, 1.0)

for i, (bar, acc) in enumerate(zip(bars, results_df['Accuracy'])):
    plt.text(acc + 0.01, i, f'{acc:.3f}', va='center', fontsize=10)
    if 'Aug' in results_df['Model'].iloc[i]:
        bar.set_edgecolor('red')
        bar.set_linewidth(2)

plt.gca().invert_yaxis()
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# ==================== FINAL REPORT ====================
print("\n" + "="*80)
print("FINAL REPORT AND RECOMMENDATIONS")
print("="*80)

best_model_row = results_df.iloc[0]
print(f"\n🎯 BEST PERFORMING MODEL: {best_model_row['Model']}")
print(f"   Accuracy: {best_model_row['Accuracy']:.4f}")
print(f"   F1-Score: {best_model_row['F1-Score']:.4f}")
print(f"   ROC AUC: {best_model_row['ROC AUC']:.4f}")

print("\n📊 DATASET STATISTICS:")
print(f"   Total classes: {len(label_encoder.classes_)}")
print(f"   Training samples: {len(X_spec_train)}")
print(f"   Validation samples: {len(X_spec_val)}")
print(f"   Test samples: {len(X_spec_test)}")
print(f"   Class imbalance ratio (max/min): {max(class_counts.values) / min(class_counts.values):.2f}")

print("\n🔧 AUGMENTATION EFFECTIVENESS:")
baseline_cnn_acc = results_df[results_df['Model'] == 'Spectrogram CNN']['Accuracy'].values[0] \
    if 'Spectrogram CNN' in results_df['Model'].values else 0.4

print(f"   Baseline CNN Accuracy: {baseline_cnn_acc:.4f}")

if 'CNN with VAE Augmentation' in results_df['Model'].values:
    vae_acc = results_df[results_df['Model'] == 'CNN with VAE Augmentation']['Accuracy'].values[0]
    vae_improvement = ((vae_acc - baseline_cnn_acc) / baseline_cnn_acc * 100) if baseline_cnn_acc > 0 else 0
    print(f"   VAE Augmentation Accuracy: {vae_acc:.4f} (Improvement: {vae_improvement:+.1f}%)")

if 'CNN with MixUp Augmentation' in results_df['Model'].values:
    mixup_acc = results_df[results_df['Model'] == 'CNN with MixUp Augmentation']['Accuracy'].values[0]
    mixup_improvement = ((mixup_acc - baseline_cnn_acc) / baseline_cnn_acc * 100) if baseline_cnn_acc > 0 else 0
    print(f"   MixUp Augmentation Accuracy: {mixup_acc:.4f} (Improvement: {mixup_improvement:+.1f}%)")


# Save results
try:
    results_df.to_csv('/content/drive/MyDrive/ocean_sound_results.csv', index=False)
    print("\n💾 Results saved to Google Drive")
except:
    results_df.to_csv('ocean_sound_results.csv', index=False)
    print("\n💾 Results saved locally")

```
